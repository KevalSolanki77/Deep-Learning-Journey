```python
import tensorflow as tf
from keras.datasets import imdb
from keras import Sequential
from keras.preprocessing.sequence import pad_sequences
from keras.layers import Input, Dense, Embedding, SimpleRNN, LSTM, GRU, Bidirectional
```

```python
num_words = 10000
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=num_words)
MAX_LEN = 100
X_train = pad_sequences(X_train, maxlen=MAX_LEN, padding = 'pre')
X_test = pad_sequences(X_test, maxlen=MAX_LEN, padding = 'pre')
```

# Deep RNN

```python
EMBEDDING_DIM = 32
UNITS = 8
```

```python
model = Sequential([
    Input(shape = (X_train.shape[1],)),
    Embedding(
        input_dim = num_words,
        output_dim = EMBEDDING_DIM
    ),
    SimpleRNN(UNITS, return_sequences=True),
    SimpleRNN(UNITS),
    Dense(1, activation ='sigmoid')
])

model.summary()

model.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics=['accuracy'])

model.fit(X_train, y_train, epochs = 5, validation_data = (X_test, y_test))
```

# Deep LSTM

```python
model = Sequential([
    Input(shape = (X_train.shape[1],)),
    Embedding(
        input_dim = num_words,
        output_dim = EMBEDDING_DIM
    ),
    LSTM(UNITS, return_sequences=True),
    LSTM(UNITS),
    Dense(1, activation ='sigmoid')
])

model.summary()

model.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics=['accuracy'])

model.fit(X_train, y_train, epochs = 5, validation_data = (X_test, y_test))
```

# Deep GRU

```python
model = Sequential([
    Input(shape = (X_train.shape[1],)),
    Embedding(
        input_dim = num_words,
        output_dim = EMBEDDING_DIM
    ),
    GRU(UNITS, return_sequences=True),
    GRU(UNITS),
    Dense(1, activation ='sigmoid')
])

model.summary()

model.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics=['accuracy'])

model.fit(X_train, y_train, epochs = 5, validation_data = (X_test, y_test))
```

# Bidirectional RNN

```python
model = Sequential([
    Input(shape = (X_train.shape[1],)),
    Embedding(
        input_dim = num_words,
        output_dim = EMBEDDING_DIM
    ),
    Bidirectional(SimpleRNN(UNITS)),
    Dense(1, activation ='sigmoid')
])

model.summary()

model.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics=['accuracy'])

model.fit(X_train, y_train, epochs = 5, validation_data = (X_test, y_test))
```

# Bidirectional LSTM

```python
model = Sequential([
    Input(shape = (X_train.shape[1],)),
    Embedding(
        input_dim = num_words,
        output_dim = EMBEDDING_DIM
    ),
    Bidirectional(LSTM(UNITS)),
    Dense(1, activation ='sigmoid')
])

model.summary()

model.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics=['accuracy'])

model.fit(X_train, y_train, epochs = 5, validation_data = (X_test, y_test))
```

# Bidirectional GRU

```python
model = Sequential([
    Input(shape = (X_train.shape[1],)),
    Embedding(
        input_dim = num_words,
        output_dim = EMBEDDING_DIM
    ),
    Bidirectional(GRU(UNITS)),
    Dense(1, activation ='sigmoid')
])

model.summary()

model.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics=['accuracy'])

model.fit(X_train, y_train, epochs = 5, validation_data = (X_test, y_test))
```

# Deep Bidirectional NN

```python
from keras.regularizers import l1_l2

model = Sequential([
    Input(shape = (X_train.shape[1],)),
    Embedding(
        input_dim = num_words,
        output_dim = EMBEDDING_DIM
    ),
    Bidirectional(GRU(UNITS, return_sequences=True, recurrent_dropout=0.2, kernel_regularizer=l1_l2(l1=0.01, l2=0.01))),
    Bidirectional(GRU(UNITS, return_sequences=True, recurrent_dropout=0.2, kernel_regularizer=l1_l2(l1=0.01, l2=0.01))),
    Bidirectional(GRU(UNITS)),
    Dense(1, activation ='sigmoid')
])

model.summary()

model.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics=['accuracy'])

model.fit(X_train, y_train, epochs = 5, validation_data = (X_test, y_test))
```
