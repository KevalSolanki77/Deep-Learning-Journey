```python
import tensorflow
from keras.layers import Input, Dense,Conv2D,Flatten
from keras import Sequential
```

# Padding

### 1. `valid` : Image Gets small after each Convolution Layer 
### 2. `same` : Image stays same after each Convolution Layer 

```python
model = Sequential()

model.add(Input(shape = (28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

```python
model = Sequential()

model.add(Input(shape = (28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

# Strides

### `(right_jump, down_jump)` : For Ex. strides = (2,2) skips every other one, so the output is roughly half the height and width.

```python
model = Sequential()

model.add(Input(shape = (28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(1,1), activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(1,1), activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(1,1), activation='relu'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

```python
model = Sequential()

model.add(Input(shape = (28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```
