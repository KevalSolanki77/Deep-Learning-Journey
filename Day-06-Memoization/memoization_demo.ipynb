# Without Memoization

```python
import time
def fib(n):
    if n == 0 or n == 1:
        return 1
    else:
        return fib(n-1) + fib(n-2)

start = time.time()
fib(40)
print(f"Time Taken : {time.time() - start}")
```

# With Memoization

```python
def fib(n, d):
    if n in d:
        return d[n]
    else:
        d[n] = fib(n-1, d) + fib(n-2, d)
        return d[n]

d = {0: 1, 1: 1}
start = time.time()
fib(40, d)
print(f"Time Taken : {time.time() - start}")
```

### So it take even less than 1 sec to calculate with Memoization.
