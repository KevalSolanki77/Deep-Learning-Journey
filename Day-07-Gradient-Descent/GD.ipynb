# Batch Vs. Stocastic GD

# Batch GD :
### Do One update after per epoch, using whole dataset

# Stocastic GD :
### Do one update per sample

```python
import numpy as np
import pandas as pd
import time

from sklearn.datasets import fetch_california_housing
```

```python
df = fetch_california_housing(as_frame=True)['frame']

df.head()
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df.drop(columns = ['MedHouseVal']), df['MedHouseVal'], test_size=0.2, random_state=0)

X_train.shape, y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print(X_train)
```

```python
import tensorflow as tf
import keras
from keras.models import Sequential
from keras.layers import Input, Dense
```

```python
batch = Sequential()

batch.add(Input(shape = (X_train.shape[1],)))
batch.add(Dense(32, activation='relu'))
batch.add(Dense(16, activation='relu'))
batch.add(Dense(1))

batch.compile(loss='mse')

start = time.time()
batch.fit(X_train, y_train, epochs=10, batch_size=len(X_train)) # batch_size = X_train.shape[0] For Batch GD means do update after whole dataset
end = time.time()

batch_time = end - start

stocastic = Sequential()

stocastic.add(Input(shape = (X_train.shape[1],)))
stocastic.add(Dense(32, activation='relu'))
stocastic.add(Dense(16, activation='relu'))
stocastic.add(Dense(1))

stocastic.compile(loss='mse')

start = time.time()
stocastic.fit(X_train, y_train, epochs=1, batch_size=1) # batch_size = 1 For Stocastic GD means do update after one sample
end = time.time()

stocastic_time = end - start

print(f"Training Time for Batch GD : {batch_time}")
print(f"Training Time for Stocastic GD : {stocastic_time}")
```

### Batch GD is faster than Stocastic in training.

# Which is faster to converge given the same #epochs?

```python
batch = Sequential()

batch.add(Input(shape = (X_train.shape[1],)))
batch.add(Dense(32, activation='relu'))
batch.add(Dense(16, activation='relu'))
batch.add(Dense(1))

batch.compile( loss='mse')

start = time.time()
batch_history = batch.fit(X_train, y_train, epochs=3, batch_size=len(X_train)) # batch_size = X_train.shape[0] For Batch GD means do update after whole dataset
end = time.time()

batch_time = end - start

stocastic = Sequential()

stocastic.add(Input(shape = (X_train.shape[1],)))
stocastic.add(Dense(32, activation='relu'))
stocastic.add(Dense(16, activation='relu'))
stocastic.add(Dense(1))

stocastic.compile(loss='mse')

start = time.time()
stocastic_history = stocastic.fit(X_train, y_train, epochs=3, batch_size=1) # batch_size = 1 For Stocastic GD means do update after one sample
end = time.time()

stocastic_time = end - start

print(f"Training Time for Batch GD : {batch_time}")
print(f"Training Time for Stocastic GD : {stocastic_time}")
```

### Batch GD has large loss than Stocastic GD so we can say that Stocastic will converge fast on the same #epochs.

### There are many downsides of these both algorithms 

### For Batch GD:
1. Takes time to converge
2. Sometimes we cannot load whole dataset in the RAM.


### For stocastic:
1. No guarantee of Best solution.
2. Sometime we get the local solution.

### So to overcome these problems we have mini-batch GD. It is the middle groound of Batch and stocastic GD.

# Mini Batch GD :

### In this version of GD we split the data into mini batches rather than full batch like Batch GD and then perform the update on these mini batches.

```python
mini_batch = Sequential()

mini_batch.add(Input(shape = (X_train.shape[1],)))
mini_batch.add(Dense(32, activation='relu'))
mini_batch.add(Dense(16, activation='relu'))
mini_batch.add(Dense(1))

mini_batch.compile( loss='mse')

start = time.time()
batch_history = mini_batch.fit(X_train, y_train, epochs=10, batch_size=512) # batch_size = X_train.shape[0] For Batch GD means do update after whole dataset
end = time.time()

mini_batch_time = end - start

print(f"Training Time for Mini-Batch GD : {mini_batch_time}")
```
