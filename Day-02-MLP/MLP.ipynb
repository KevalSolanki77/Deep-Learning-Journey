```python
import numpy as np
import pandas as pd
```

```python
train_df = pd.read_csv("data/customer_churn_dataset-training-master.csv")
train_df.shape
```

```python
train_df.head()
```

```python
train_df.info()
```

```python
train_df.duplicated().sum()
```

```python
train_df.isnull().sum()
```

```python
train_df.dropna(inplace=True)
```

```python
train_df.drop(columns = ['CustomerID'], inplace = True)
```

```python
train_df['Subscription Type'].value_counts()
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(categories=[['Basic', 'Standard', 'Premium']])

train_df['Subscription Type'] = encoder.fit_transform(train_df[['Subscription Type']])
train_df = pd.get_dummies(data = train_df, columns = ['Contract Length', 'Gender'], drop_first = True, dtype=int)
train_df.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(train_df.drop(columns = ['Churn']), train_df['Churn'], test_size = 0.3, random_state = 0)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns = scaler.get_feature_names_out())
X_val_scaled = pd.DataFrame(scaler.transform(X_val), columns = scaler.get_feature_names_out())
```

```python
import tensorflow as tf
from tensorflow import keras

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense

model = Sequential()

model.add(Dense(3, activation='sigmoid', input_dim = 11))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(loss = 'binary_crossentropy', optimizer='Adam')
```

```python
model.fit(X_train_scaled, y_train, epochs=10)
```

```python
model.layers[0].get_weights() # Weights and biases of the input layer
```

```python
y_prob = model.predict(X_val_scaled)

threshold = 0.8

y_hat = np.where(y_prob > threshold, 1, 0)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_val, y_hat)
```
